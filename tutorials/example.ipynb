```python
from typing import List, Optional, Tuple
import cv2
import numpy as np

def get_inner_pericarp_area(
    locules: List[int], 
    contours: List[np.ndarray], 
    px_per_cm: Optional[float] = None, 
    img: Optional[np.ndarray] = None,
    draw_inner_pericarp: Optional[bool] = False, 
    use_ellipse: bool = False, 
    contour_thickness: int = 2, 
    contour_color: Tuple[int, int, int] = (0, 240, 240)
) -> Tuple[float, float, np.ndarray]:
    if draw_inner_pericarp and img is None:
        raise ValueError("img cannot be None when draw_inner_pericarp=True")

    if not locules:
        return 0.0, 0.0, None

    all_points = np.vstack([contours[i] for i in locules])
    area_px = 0.0

    # máscara binaria (una sola vez)
    if img is not None:
        mask = np.zeros(img.shape[:2], dtype=np.uint8)
    else:
        # si no te pasan img, usa el bounding box de los puntos
        x, y, w, h = cv2.boundingRect(all_points)
        mask = np.zeros((y + h, x + w), dtype=np.uint8)

    smoothed_hull = cv2.convexHull(all_points)
    #epsilon_val = 0.0001 * cv2.arcLength(hull, True)
    #smoothed_hull = cv2.approxPolyDP(hull, epsilon_val, True)
    # Draw yellow line around all the locules
    cv2.drawContours(img, [smoothed_hull], -1, (contour_color), contour_thickness)
    
    
    if draw_inner_pericarp and img is not None:
        cv2.drawContours(mask, [smoothed_hull], -1, 255, -1)

    area_px = cv2.contourArea(smoothed_hull)

    if px_per_cm is not None and not isinstance(px_per_cm, dict) and isinstance(px_per_cm, (int, float)) and px_per_cm > 0:
        area_cm2 = area_px / (px_per_cm ** 2)
    else:
        area_cm2 = np.nan
    if draw_inner_pericarp:
        return area_cm2, area_px, mask 
    else:
        return area_cm2, area_px

```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

# ============================
# 1. Crear contorno de flor de 4 pétalos
# ============================

h, w = 400, 400
img = np.zeros((h, w, 3), dtype=np.uint8)

# Centro de la flor
cx, cy = w // 2, h // 2

# Parámetros de la flor
a = 120  # "radio" máximo de la flor
theta = np.linspace(0, 2 * np.pi, 800)

# Curva tipo "rose": r = a * cos(2θ)
r = a * np.cos(2 * theta)

# Usamos sólo puntos donde r > 0 (4 pétalos)
mask = r > 0
r = r[mask]
theta_used = theta[mask]

x = cx + r * np.cos(theta_used)
y = cy + r * np.sin(theta_used)

# Convertir a contorno para OpenCV (int32, forma Nx1x2)
contour = np.stack([x, y], axis=1).astype(np.int32).reshape(-1, 1, 2)

# Dibujar contorno original
img_contour = img.copy()
cv2.drawContours(img_contour, [contour], -1, (0, 255, 255), 2)  # amarillo


# ============================
# 2. Convex hull
# ============================

hull = cv2.convexHull(contour)

img_hull = img.copy()
cv2.drawContours(img_hull, [hull], -1, (0, 255, 0), 2)  # verde


# ============================
# 3. Aproximación poligonal del hull
# ============================

epsilon = 0.02 * cv2.arcLength(hull, True)  # grado de simplificación
poly = cv2.approxPolyDP(hull, epsilon, True)

img_poly = img.copy()
cv2.drawContours(img_poly, [hull], -1, (0, 255, 0), 2)   # hull en verde
cv2.drawContours(img_poly, [poly], -1, (255, 0, 0), 2)   # polígono en azul


# ============================
# 4. Mostrar resultados
# ============================

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(cv2.cvtColor(img_contour, cv2.COLOR_BGR2RGB))
axes[0].set_title("Contorno original (flor 4 pétalos)")
axes[0].axis("off")

axes[1].imshow(cv2.cvtColor(img_hull, cv2.COLOR_BGR2RGB))
axes[1].set_title("Convex hull")
axes[1].axis("off")

axes[2].imshow(cv2.cvtColor(img_poly, cv2.COLOR_BGR2RGB))
axes[2].set_title("Hull (verde) + Polígono (azul)")
axes[2].axis("off")

plt.tight_layout()
plt.show()

# ============================
# 5. Áreas (opcional, por curiosidad)
# ============================

area_hull = cv2.contourArea(hull)
area_poly = cv2.contourArea(poly)
print("Área hull      :", area_hull)
print("Área polígono :", area_poly)

```

```python
from dataclasses import dataclass
from typing import Tuple, Optional, List
import cv2
import numpy as np

def draw_pericarp_rays(
    img: np.ndarray,
    outer_contour: np.ndarray,
    inner_contour: np.ndarray,
    fruit_centroid: Tuple[int, int],
    num_rays: int = 180,
    ray_color: Tuple[int, int, int] = (0, 0, 0),  # Cyan
    outer_color: Tuple[int, int, int] = (0, 255, 0),  # Green
    inner_color: Tuple[int, int, int] = (255, 0, 255),  # Blue
    ray_thickness: int = 1,
    contour_thickness: int = 2
):
    """
    Draw radial rays and contours for pericarp thickness visualization.
    Uses the SAME intersection detection logic as calculate_pericarp_thickness_radial.
    
    Args:
        img: Image to draw on (will be modified in place)
        outer_contour: Outer fruit contour
        inner_contour: Inner contour (convex hull of locules)
        fruit_centroid: Fruit center (cx, cy)
        num_rays: Number of rays to draw
        ray_color: Color for rays (BGR)
        outer_color: Color for outer contour (BGR)
        inner_color: Color for inner contour (BGR)
        ray_thickness: Thickness of ray lines
        contour_thickness: Thickness of contour lines
    """
    cx, cy = fruit_centroid
    height, width = img.shape[:2]
    
    # Get bounding box to create smaller masks
    x_out, y_out, w_out, h_out = cv2.boundingRect(outer_contour)
    
    # Add margin for safety
    margin = 10
    x_min = max(0, x_out - margin)
    y_min = max(0, y_out - margin)
    x_max = min(width, x_out + w_out + margin)
    y_max = min(height, y_out + h_out + margin)
    
    # Create smaller masks (only region of interest)
    mask_width = x_max - x_min
    mask_height = y_max - y_min
    
    # ✅ FIX: Use uint8 instead of bool for drawContours
    mask_outer = np.zeros((mask_height, mask_width), dtype=np.uint8)
    mask_inner = np.zeros((mask_height, mask_width), dtype=np.uint8)
    
    # Adjust contours to local coordinates
    outer_local = outer_contour - np.array([x_min, y_min])
    inner_local = inner_contour - np.array([x_min, y_min])
    
    # ✅ FIX: Use 255 instead of True
    cv2.drawContours(mask_outer, [outer_local], -1, 255, -1)
    if len(inner_contour) > 0:  # Only draw if inner contour exists
        cv2.drawContours(mask_inner, [inner_local], -1, 255, -1)
    
    # Adjust centroid to local coordinates
    cx_local = cx - x_min
    cy_local = cy - y_min
    
    # Precalculate all rays
    angles = np.linspace(0, 2 * np.pi, num_rays, endpoint=False)
    max_search = int(np.hypot(mask_width, mask_height))
    
    # Create radial distance grid
    r_grid = np.arange(1, max_search)
    cos_angles = np.cos(angles)
    sin_angles = np.sin(angles)
    
    # Process rays and draw them
    for dx, dy in zip(cos_angles, sin_angles):
        # Calculate all positions along the ray
        xs = (cx_local + dx * r_grid).astype(int)
        ys = (cy_local + dy * r_grid).astype(int)
        
        # Filter valid coordinates
        valid = (xs >= 0) & (xs < mask_width) & (ys >= 0) & (ys < mask_height)
        xs_valid = xs[valid]
        ys_valid = ys[valid]
        r_valid = r_grid[valid]
        
        if len(xs_valid) == 0:
            continue
        
        # Find outer boundary intersection
        outer_vals = mask_outer[ys_valid, xs_valid]
        # ✅ FIX: Compare with 0 instead of using ~ (because uint8, not bool)
        outer_idx = np.where(outer_vals == 0)[0]
        
        if len(outer_idx) == 0:
            continue
        
        outer_r = r_valid[outer_idx[0] - 1] if outer_idx[0] > 0 else 0
        
        # Calculate END POINT of ray at outer boundary
        end_x_local = int(cx_local + dx * outer_r)
        end_y_local = int(cy_local + dy * outer_r)
        
        # Convert back to global coordinates
        end_x = end_x_local + x_min
        end_y = end_y_local + y_min
        
        # Draw ray from centroid to outer boundary (not beyond!)
        cv2.line(img, (cx, cy), (end_x, end_y), ray_color, ray_thickness)
    
    # Draw contours
    cv2.drawContours(img, [outer_contour], -1, outer_color, contour_thickness)
    if len(inner_contour) > 0:  # Only draw if inner contour exists
        cv2.drawContours(img, [inner_contour], -1, inner_color, contour_thickness)
    
    # Draw centroid
    cv2.circle(img, (cx, cy), 5, (255, 255, 0), -1)  # White dot
    cv2.circle(img, (cx, cy), 5, (0, 0, 0), 1)  # Black outline

@dataclass
class PericarpVisualizationConfig:
    """Configuration for pericarp visualization (no calculations, just drawing)."""
    # Ray settings
    num_rays: int = 180
    ray_color: Tuple[int, int, int] = (255,255, 255)  # Cyan
    ray_thickness: int = 1
    draw_rays: bool = True
    
    # Contour settings
    outer_contour_color: Tuple[int, int, int] = (0, 255, 0)  # Green
    inner_contour_color: Tuple[int, int, int] = (255, 0, 255)  # Blue
    contour_thickness: int = 2
    draw_outer_contour: bool = True
    draw_inner_contour: bool = True
    
    # Centroid settings
    centroid_color: Tuple[int, int, int] = (255, 0, 255)  # White
    centroid_outline_color: Tuple[int, int, int] = (0, 0, 0)  # Black
    centroid_radius: int = 5
    draw_centroid: bool = True
    
    # Optional: Draw locule contours individually
    draw_locule_contours: bool = False
    locule_contour_color: Tuple[int, int, int] = (255, 165, 0)  # Orange
    locule_contour_thickness: int = 1

def visualize_pericarp_structure(
    img: np.ndarray,
    fruit_contour: np.ndarray,
    locule_indices: List[int],
    contours: List[np.ndarray],
    fruit_centroid: Tuple[int, int],
    config: Optional[PericarpVisualizationConfig] = None
) -> np.ndarray:
    """Visualize pericarp with accurate ray intersections."""
    
    if config is None:
        config = PericarpVisualizationConfig()
    
    vis_img = img.copy()
    cx, cy = fruit_centroid
    
    # Get inner contour if needed
    if config.draw_inner_contour and locule_indices:
        locule_contours = [contours[i] for i in locule_indices]
        all_locule_points = np.vstack(locule_contours)
        inner_contour = cv2.convexHull(all_locule_points)
    else:
        inner_contour = None
    
    # ✅ Use the accurate ray drawing function
    if config.draw_rays or config.draw_outer_contour or config.draw_inner_contour or config.draw_centroid:
        # Prepare inner contour (empty array if not available)
        inner_for_draw = inner_contour if inner_contour is not None else np.array([])
        
        draw_pericarp_rays(
            img=vis_img,
            outer_contour=fruit_contour,
            inner_contour=inner_for_draw,
            fruit_centroid=fruit_centroid,
            num_rays=config.num_rays,
            ray_color=config.ray_color,
            outer_color=config.outer_contour_color,
            inner_color=config.inner_contour_color,
            ray_thickness=config.ray_thickness,
            contour_thickness=config.contour_thickness
        )
    
    # Draw individual locules if requested
    if config.draw_locule_contours and locule_indices:
        for locule_id in locule_indices:
            cv2.drawContours(vis_img, [contours[locule_id]], -1,
                           config.locule_contour_color, config.locule_contour_thickness)
    
    return vis_img


def quick_visualize_pericarp(
    img: np.ndarray,
    fruit_contour: np.ndarray,
    locule_indices: List[int],
    contours: List[np.ndarray],
    fruit_centroid: Tuple[int, int],
    num_rays: int = 180
) -> np.ndarray:
    """
    Quick visualization with default settings.
    
    Args:
        img: Input image
        fruit_contour: Outer fruit contour
        locule_indices: Indices of locule contours
        contours: List of all contours
        fruit_centroid: Fruit center (cx, cy)
        num_rays: Number of rays to draw
    
    Returns:
        Annotated image
    """
    config = PericarpVisualizationConfig(num_rays=num_rays)
    return visualize_pericarp_structure(
        img, fruit_contour, locule_indices, contours, fruit_centroid, config
    )


def visualize_all_fruits_pericarp(
    analyzer,
    num_rays: int = 180,
    save_individual: bool = False,
    output_prefix: str = "fruit"
) -> np.ndarray:
    """
    Convenience function to visualize all fruits in an analyzer.
    
    Args:
        analyzer: FruitAnalyzer object with fruit_locus_map and contours
        num_rays: Number of rays to draw per fruit
        save_individual: If True, save each fruit as separate image
        output_prefix: Prefix for saved images
    
    Returns:
        Image with all fruits visualized
    """
    result_img = analyzer.img.copy()
    
    for fruit_id, locules in analyzer.fruit_locus_map.items():
        fruit_contour = analyzer.contours[fruit_id]
        
        # Calculate centroid
        M = cv2.moments(fruit_contour)
        if M["m00"] == 0:
            continue
        
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
        fruit_centroid = (cx, cy)
        
        # Visualize this fruit
        if save_individual:
            # Create separate image for this fruit
            individual_img = quick_visualize_pericarp(
                img=analyzer.img.copy(),
                fruit_contour=fruit_contour,
                locule_indices=locules,
                contours=analyzer.contours,
                fruit_centroid=fruit_centroid,
                num_rays=num_rays
            )
            cv2.imwrite(f"{output_prefix}_{fruit_id}_rays.png", individual_img)
        
        # Add to combined image
        result_img = quick_visualize_pericarp(
            img=result_img,
            fruit_contour=fruit_contour,
            locule_indices=locules,
            contours=analyzer.contours,
            fruit_centroid=fruit_centroid,
            num_rays=num_rays
        )
    
    return result_img
```

```python
def get_fruit_contour(
    img: Optional[np.array],
    contours: List[np.ndarray], 
    fruit_id: int, 
    contour_mode: str = 'raw', 
    epsilon_factor: float = 0.0001
) -> np.ndarray:
    """
    Extract and optionally transform a fruit contour.
    
    Args:
        contours: List of contours from cv2.findContours
        fruit_id: Index of the fruit contour to extract
        contour_mode: Transformation mode:
            - 'raw': Original contour (default)
            - 'hull': Convex hull approximation
            - 'approx': Approximate polygon using Douglas-Peucker
            - 'ellipse': Fit an ellipse around the contour
            - 'circle': Fit minimum enclosing circle
        epsilon_factor: Approximation factor for 'approx' mode
        
    Returns:
        Transformed contour points
    """
    valid_modes = ['raw', 'hull', 'approx', 'ellipse', 'circle']
    if contour_mode not in valid_modes:
        raise ValueError(f"contour_mode must be one of {valid_modes}, got '{contour_mode}'")
    
    if not 0 <= fruit_id < len(contours):
        raise IndexError(f"fruit_id {fruit_id} out of range [0, {len(contours)-1}]")
    
    fruit_contour = contours[fruit_id]
    
    if contour_mode == 'hull':
        fruit_contour = cv2.convexHull(fruit_contour)
        
    elif contour_mode == 'approx':
        peri = cv2.arcLength(fruit_contour, True)
        epsilon = epsilon_factor * peri
        fruit_contour = cv2.approxPolyDP(fruit_contour, epsilon, True)
        
    elif contour_mode == 'ellipse':
        if len(fruit_contour) < 5:
            raise ValueError("Ellipse fitting requires at least 5 contour points")
        ellipse = cv2.fitEllipse(fruit_contour)
        center = (int(ellipse[0][0]), int(ellipse[0][1]))
        axes = (int(ellipse[1][0] / 2), int(ellipse[1][1] / 2))
        angle = int(ellipse[2])
        fruit_contour = cv2.ellipse2Poly(center, axes, angle, 0, 360, 2).reshape(-1, 1, 2)
    
    elif contour_mode == 'circle':
        (x, y), radius = cv2.minEnclosingCircle(fruit_contour)
        center = (int(x), int(y))
        radius = int(radius)
        
        angles = np.linspace(0, 2 * np.pi, 36, endpoint=False)
        circle_points = np.column_stack([
            center[0] + radius * np.cos(angles),
            center[1] + radius * np.sin(angles)
        ]).astype(np.int32)
        
        fruit_contour = circle_points.reshape(-1, 1, 2)

    cv2.drawContours(img, [fruit_contour], -1, (0, 255, 0), 2)
    
    return None
```

```python
from dataclasses import dataclass
from typing import Tuple, Optional, List, Dict
import cv2
import numpy as np

@dataclass
class DebugVisualizationConfig:
    """Configuration for step-by-step debugging visualization."""
    # Which steps to visualize
    show_bounding_box: bool = True
    show_masks: bool = True
    show_all_rays: bool = True
    show_clipped_rays: bool = True
    show_outer_intersections: bool = True
    show_inner_intersections: bool = True
    show_final_result: bool = True
    
    # Visual settings
    num_rays: int = 45
    ray_color: Tuple[int, int, int] = (255, 255, 255)
    ray_thickness: int = 1
    bbox_color: Tuple[int, int, int] = (255, 0, 255)  # Magenta
    bbox_thickness: int = 2
    
    # Contour colors
    outer_contour_color: Tuple[int, int, int] = (0, 255, 0)  # Green
    inner_contour_color: Tuple[int, int, int] = (255, 0, 255)  # Blue
    contour_thickness: int = 2
    
    # Intersection point colors
    outer_point_color: Tuple[int, int, int] = (0, 255, 0)  # Green
    inner_point_color: Tuple[int, int, int] = (255, 0, 0)  # Blue
    point_radius: int = 3


def visualize_pericarp_step_by_step(
    img: np.ndarray,
    fruit_contour: np.ndarray,
    locule_indices: List[int],
    contours: List[np.ndarray],
    fruit_centroid: Tuple[int, int],
    config: Optional[DebugVisualizationConfig] = None
) -> Dict[str, np.ndarray]:
    """
    Generate step-by-step visualization of pericarp thickness calculation.
    USES CORRECTED INTERSECTION LOGIC.
    
    Returns a dictionary with images for each step.
    """
    if config is None:
        config = DebugVisualizationConfig()
    
    cx, cy = fruit_centroid
    height, width = img.shape[:2]
    
    result_images = {}
    
    # ============ STEP 1: BOUNDING BOX ============
    if config.show_bounding_box:
        step1_img = img.copy()
        x_out, y_out, w_out, h_out = cv2.boundingRect(fruit_contour)
        
        # Draw original bounding box (tight)
        cv2.rectangle(step1_img, (x_out, y_out), 
                     (x_out + w_out, y_out + h_out), 
                     (0, 0, 255), 2)  # Red for tight bbox
        
        # Add margin
        margin = 10
        x_min = max(0, x_out - margin)
        y_min = max(0, y_out - margin)
        x_max = min(width, x_out + w_out + margin)
        y_max = min(height, y_out + h_out + margin)
        
        # Draw expanded bounding box
        cv2.rectangle(step1_img, (x_min, y_min), (x_max, y_max), 
                     config.bbox_color, config.bbox_thickness)
        
        # Draw fruit contour
        cv2.drawContours(step1_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        
        # Draw centroid
        cv2.circle(step1_img, (cx, cy), 8, (255, 255, 0), -1)
        cv2.circle(step1_img, (cx, cy), 8, (0, 0, 0), 2)
        
        # Add text annotation
        cv2.putText(step1_img, "Step 1: Bounding Box", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        cv2.putText(step1_img, f"Red: Tight bbox | Magenta: +{margin}px margin", 
                   (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)
        
        result_images['step1_bounding_box'] = step1_img
    else:
        # Still need to calculate for next steps
        x_out, y_out, w_out, h_out = cv2.boundingRect(fruit_contour)
        margin = 10
        x_min = max(0, x_out - margin)
        y_min = max(0, y_out - margin)
        x_max = min(width, x_out + w_out + margin)
        y_max = min(height, y_out + h_out + margin)
    
    # ============ STEP 2: MASKS ============
    mask_width = x_max - x_min
    mask_height = y_max - y_min
    
    # Create masks
    mask_outer = np.zeros((mask_height, mask_width), dtype=np.uint8)
    mask_inner = np.zeros((mask_height, mask_width), dtype=np.uint8)
    
    # Adjust contours to local coordinates
    outer_local = fruit_contour - np.array([x_min, y_min])
    
    # Get inner contour
    if locule_indices:
        locule_contours = [contours[i] for i in locule_indices]
        all_locule_points = np.vstack(locule_contours)
        inner_contour = cv2.convexHull(all_locule_points)
        inner_local = inner_contour - np.array([x_min, y_min])
    else:
        inner_contour = np.array([])
        inner_local = np.array([])
    
    # Fill masks
    cv2.drawContours(mask_outer, [outer_local], -1, 255, -1)
    if len(inner_local) > 0:
        cv2.drawContours(mask_inner, [inner_local], -1, 255, -1)
    
    if config.show_masks:
        # Create visualization: show both masks side by side
        step2_img = np.zeros((mask_height, mask_width * 2 + 20, 3), dtype=np.uint8)
        
        # Left: outer mask (green channel)
        step2_img[:, :mask_width, 1] = mask_outer
        
        # Right: inner mask (blue channel)
        step2_img[:, mask_width + 20:, 0] = mask_inner
        
        # Add labels
        cv2.putText(step2_img, "Outer Mask", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)
        cv2.putText(step2_img, "Inner Mask", (mask_width + 30, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 0, 0), 2)
        
        result_images['step2_masks'] = step2_img
    
    # Adjust centroid to local coordinates
    cx_local = cx - x_min
    cy_local = cy - y_min
    
    # Precalculate rays
    angles = np.linspace(0, 2 * np.pi, config.num_rays, endpoint=False)
    max_search = int(np.hypot(mask_width, mask_height))
    r_grid = np.arange(1, max_search)
    cos_angles = np.cos(angles)
    sin_angles = np.sin(angles)
    
    # ============ STEP 3: ALL RAYS (BEFORE CLIPPING) ============
    if config.show_all_rays:
        step3_img = img.copy()
        
        # Draw bounding box for reference
        cv2.rectangle(step3_img, (x_min, y_min), (x_max, y_max), 
                     config.bbox_color, 1)
        
        for dx, dy in zip(cos_angles, sin_angles):
            # Draw ray to maximum theoretical distance
            end_x = int(cx + dx * max_search)
            end_y = int(cy + dy * max_search)
            cv2.line(step3_img, (cx, cy), (end_x, end_y), 
                    (100, 100, 100), 1)  # Gray for unclipped rays
        
        # Draw contours
        cv2.drawContours(step3_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        cv2.circle(step3_img, (cx, cy), 5, (255, 255, 0), -1)
        
        cv2.putText(step3_img, "Step 3: All Rays (unclipped)", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        cv2.putText(step3_img, f"Max distance: {max_search}px (diagonal)", 
                   (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)
        
        result_images['step3_all_rays'] = step3_img
    
    # ============ STEP 4: CLIPPED TO BOUNDING BOX ============
    if config.show_clipped_rays:
        step4_img = img.copy()
        
        # Draw bounding box
        cv2.rectangle(step4_img, (x_min, y_min), (x_max, y_max), 
                     config.bbox_color, config.bbox_thickness)
        
        for dx, dy in zip(cos_angles, sin_angles):
            xs = (cx_local + dx * r_grid).astype(int)
            ys = (cy_local + dy * r_grid).astype(int)
            
            # Find where ray exits bounding box
            valid = (xs >= 0) & (xs < mask_width) & (ys >= 0) & (ys < mask_height)
            
            if not valid.any():
                continue
            
            # Get last valid point
            last_valid_idx = np.where(valid)[0][-1]
            end_x_local = xs[last_valid_idx]
            end_y_local = ys[last_valid_idx]
            
            # Convert to global coordinates
            end_x = end_x_local + x_min
            end_y = end_y_local + y_min
            
            cv2.line(step4_img, (cx, cy), (end_x, end_y), 
                    config.ray_color, config.ray_thickness)
        
        cv2.drawContours(step4_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        cv2.circle(step4_img, (cx, cy), 5, (255, 255, 0), -1)
        
        cv2.putText(step4_img, "Step 4: Rays Clipped to BBox", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        result_images['step4_clipped_rays'] = step4_img
    
    # ============ STEP 5: OUTER INTERSECTIONS (CORRECTED LOGIC) ============
    outer_intersections = []
    
    for dx, dy in zip(cos_angles, sin_angles):
        xs = (cx_local + dx * r_grid).astype(int)
        ys = (cy_local + dy * r_grid).astype(int)
        
        valid = (xs >= 0) & (xs < mask_width) & (ys >= 0) & (ys < mask_height)
        xs_valid = xs[valid]
        ys_valid = ys[valid]
        r_valid = r_grid[valid]
        
        if len(xs_valid) == 0:
            continue
        
        # CORRECTED: Find outer boundary (last pixel INSIDE mask)
        outer_vals = mask_outer[ys_valid, xs_valid]
        outer_idx = np.where(outer_vals == 0)[0]  # First pixel OUTSIDE
        
        if len(outer_idx) == 0:
            # All pixels are inside, use last valid position
            outer_r = r_valid[-1]
        elif outer_idx[0] == 0:
            # First pixel already outside, skip this ray
            continue
        else:
            # Get last pixel inside (one before first outside pixel)
            outer_r = r_valid[outer_idx[0] - 1]
        
        if outer_r > 0:
            end_x_local = int(cx_local + dx * outer_r)
            end_y_local = int(cy_local + dy * outer_r)
            outer_intersections.append((end_x_local + x_min, end_y_local + y_min))
    
    if config.show_outer_intersections:
        step5_img = img.copy()
        
        # Draw rays to outer intersections
        for end_x, end_y in outer_intersections:
            cv2.line(step5_img, (cx, cy), (end_x, end_y), 
                    config.ray_color, config.ray_thickness)
            cv2.circle(step5_img, (end_x, end_y), 
                      config.point_radius, config.outer_point_color, -1)
        
        cv2.drawContours(step5_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        cv2.circle(step5_img, (cx, cy), 5, (255, 255, 0), -1)
        
        cv2.putText(step5_img, "Step 5: Outer Contour Intersections (CORRECTED)", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        cv2.putText(step5_img, f"Found {len(outer_intersections)} intersections", 
                   (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)
        
        result_images['step5_outer_intersections'] = step5_img
    
    # ============ STEP 6: INNER INTERSECTIONS (CORRECTED LOGIC) ============
    inner_intersections = []
    thickness_lines = []
    thickness_values = []
    
    for dx, dy in zip(cos_angles, sin_angles):
        xs = (cx_local + dx * r_grid).astype(int)
        ys = (cy_local + dy * r_grid).astype(int)
        
        valid = (xs >= 0) & (xs < mask_width) & (ys >= 0) & (ys < mask_height)
        xs_valid = xs[valid]
        ys_valid = ys[valid]
        r_valid = r_grid[valid]
        
        if len(xs_valid) == 0:
            continue
        
        # CORRECTED: Outer intersection
        outer_vals = mask_outer[ys_valid, xs_valid]
        outer_idx = np.where(outer_vals == 0)[0]
        
        if len(outer_idx) == 0:
            outer_r = r_valid[-1]
        elif outer_idx[0] == 0:
            continue
        else:
            outer_r = r_valid[outer_idx[0] - 1]
        
        # Inner intersection (first pixel INSIDE inner mask)
        inner_vals = mask_inner[ys_valid, xs_valid]
        inner_idx = np.where(inner_vals == 255)[0]
        
        if len(inner_idx) == 0:
            continue
        
        inner_r = r_valid[inner_idx[0]]
        
        # Calculate thickness (using actual radial distances)
        if outer_r > inner_r:
            thickness = outer_r - inner_r
            thickness_values.append(thickness)
            
            inner_x_local = int(cx_local + dx * inner_r)
            inner_y_local = int(cy_local + dy * inner_r)
            outer_x_local = int(cx_local + dx * outer_r)
            outer_y_local = int(cy_local + dy * outer_r)
            
            inner_intersections.append((inner_x_local + x_min, inner_y_local + y_min))
            thickness_lines.append((
                (inner_x_local + x_min, inner_y_local + y_min),
                (outer_x_local + x_min, outer_y_local + y_min),
                thickness
            ))
    
    if config.show_inner_intersections:
        step6_img = img.copy()
        
        # Calculate stats for display
        if thickness_values:
            mean_thickness = np.mean(thickness_values)
            min_thickness = np.min(thickness_values)
            max_thickness = np.max(thickness_values)
        else:
            mean_thickness = min_thickness = max_thickness = 0
        
        # Draw thickness lines with color coding
        for (inner_pt, outer_pt, thickness) in thickness_lines:
            # Color based on thickness (green=thick, red=thin)
            if thickness_values:
                normalized = (thickness - min_thickness) / (max_thickness - min_thickness + 1e-6)
                color = (0, int(255 * normalized), int(255 * (1 - normalized)))
            else:
                color = (0, 255, 255)
            
            cv2.line(step6_img, inner_pt, outer_pt, color, 2)
        
        # Draw intersection points
        for inner_x, inner_y in inner_intersections:
            cv2.circle(step6_img, (inner_x, inner_y), 
                      config.point_radius, config.inner_point_color, -1)
        
        # Draw contours
        cv2.drawContours(step6_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        if len(inner_contour) > 0:
            cv2.drawContours(step6_img, [inner_contour], -1, 
                            config.inner_contour_color, config.contour_thickness)
        cv2.circle(step6_img, (cx, cy), 5, (255, 255, 0), -1)
        
        cv2.putText(step6_img, "Step 6: Thickness Measurements (CORRECTED)", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        cv2.putText(step6_img, 
                   f"Valid: {len(thickness_lines)} | Mean: {mean_thickness:.1f}px | Range: [{min_thickness:.1f}, {max_thickness:.1f}]", 
                   (10, 60), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 1)
        
        result_images['step6_inner_intersections'] = step6_img
    
    # ============ STEP 7: FINAL RESULT ============
    if config.show_final_result:
        step7_img = img.copy()
        
        # Draw all rays
        for (inner_pt, outer_pt, _) in thickness_lines:
            cv2.line(step7_img, (cx, cy), outer_pt, 
                    config.ray_color, config.ray_thickness)
        
        # Draw contours
        cv2.drawContours(step7_img, [fruit_contour], -1, 
                        config.outer_contour_color, config.contour_thickness)
        if len(inner_contour) > 0:
            cv2.drawContours(step7_img, [inner_contour], -1, 
                            config.inner_contour_color, config.contour_thickness)
        cv2.circle(step7_img, (cx, cy), 8, (255, 255, 0), -1)
        cv2.circle(step7_img, (cx, cy), 8, (0, 0, 0), 2)
        
        cv2.putText(step7_img, "Step 7: Final Visualization", (10, 30),
                   cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
        
        result_images['step7_final'] = step7_img
    
    return result_images


def visualize_all_fruits_step_by_step(
    analyzer,
    fruit_id: Optional[int] = None,
    config: Optional[DebugVisualizationConfig] = None,
    save_steps: bool = True,
    output_prefix: str = "debug"
) -> Dict[int, Dict[str, np.ndarray]]:
    """
    Generate step-by-step visualizations for one or all fruits.
    """
    if config is None:
        config = DebugVisualizationConfig()
    
    all_results = {}
    
    # Select which fruits to process
    if fruit_id is not None:
        fruit_ids = [fruit_id] if fruit_id in analyzer.fruit_locus_map else []
    else:
        fruit_ids = list(analyzer.fruit_locus_map.keys())
    
    for fid in fruit_ids:
        locules = analyzer.fruit_locus_map[fid]
        fruit_contour = analyzer.contours[fid]
        
        # Calculate centroid
        M = cv2.moments(fruit_contour)
        if M["m00"] == 0:
            continue
        
        cx = int(M["m10"] / M["m00"])
        cy = int(M["m01"] / M["m00"])
        fruit_centroid = (cx, cy)
        
        # Generate step-by-step images
        step_images = visualize_pericarp_step_by_step(
            img=analyzer.img.copy(),
            fruit_contour=fruit_contour,
            locule_indices=locules,
            contours=analyzer.contours,
            fruit_centroid=fruit_centroid,
            config=config
        )
        
        all_results[fid] = step_images
        
        # Save if requested
        if save_steps:
            for step_name, step_img in step_images.items():
                filename = f"{output_prefix}_fruit{fid}_{step_name}.png"
                cv2.imwrite(filename, step_img)
                print(f"Saved: {filename}")
    
    return all_results


# ============ CONVENIENCE FUNCTIONS ============

def quick_debug_single_fruit(
    analyzer,
    fruit_id: int,
    num_rays: int = 45,
    save_steps: bool = True
) -> Dict[str, np.ndarray]:
    """
    Quick debug visualization for a single fruit.
    
    Usage:
        steps = quick_debug_single_fruit(analyzer, fruit_id=0, num_rays=45)
        
        # Display specific step
        cv2.imshow('Outer Intersections', steps['step5_outer_intersections'])
        cv2.imshow('Thickness', steps['step6_inner_intersections'])
        cv2.waitKey(0)
    """
    config = DebugVisualizationConfig(num_rays=num_rays)
    results = visualize_all_fruits_step_by_step(
        analyzer=analyzer,
        fruit_id=fruit_id,
        config=config,
        save_steps=save_steps,
        output_prefix=f"debug_fruit{fruit_id}"
    )
    return results[fruit_id]


def quick_debug_all_fruits(
    analyzer,
    num_rays: int = 45,
    save_steps: bool = True
) -> Dict[int, Dict[str, np.ndarray]]:
    """
    Quick debug visualization for all fruits.
    
    Usage:
        all_steps = quick_debug_all_fruits(analyzer, num_rays=45)
        
        # Access specific fruit and step
        fruit0_thickness = all_steps[0]['step6_inner_intersections']
        cv2.imshow('Fruit 0 - Thickness', fruit0_thickness)
    """
    config = DebugVisualizationConfig(num_rays=num_rays)
    return visualize_all_fruits_step_by_step(
        analyzer=analyzer,
        fruit_id=None,
        config=config,
        save_steps=save_steps,
        output_prefix="debug"
    )
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from typing import Dict, Tuple, Optional


# ============ FUNCIÓN DE VALIDACIÓN ============

def ensure_uint8(L: np.ndarray) -> np.ndarray:
    """Asegura que la imagen esté en formato uint8."""
    if L.dtype != np.uint8:
        # Si está en float [0, 1], escalar a [0, 255]
        if L.max() <= 1.0:
            L = (L * 255).astype(np.uint8)
        else:
            # Si está en otro rango, clip y convertir
            L = np.clip(L, 0, 255).astype(np.uint8)
    return L


# ============ TRANSFORMACIONES LINEALES ============

def contrast_stretch(L: np.ndarray, percentile: Tuple[int, int] = (2, 98)) -> np.ndarray:
    """Contrast stretching con percentiles."""
    L = ensure_uint8(L)
    p_low, p_high = np.percentile(L, percentile)
    L_stretched = np.clip((L - p_low) / (p_high - p_low) * 255, 0, 255)
    return L_stretched.astype(np.uint8)


def linear_scaling(L: np.ndarray, alpha: float = 1.5, beta: int = 0) -> np.ndarray:
    """Escalado lineal: alpha=contraste, beta=brillo."""
    L = ensure_uint8(L)
    L_adjusted = np.clip(alpha * L + beta, 0, 255)
    return L_adjusted.astype(np.uint8)


# ============ ECUALIZACIÓN DE HISTOGRAMA ============

def histogram_equalization(L: np.ndarray) -> np.ndarray:
    """Ecualización global de histograma."""
    L = ensure_uint8(L)
    return cv2.equalizeHist(L)


def clahe_enhancement(L: np.ndarray, clip_limit: float = 2.0, 
                      tile_size: Tuple[int, int] = (8, 8)) -> np.ndarray:
    """CLAHE - Ecualización adaptativa."""
    L = ensure_uint8(L)
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_size)
    return clahe.apply(L)


# ============ TRANSFORMACIONES GAMMA ============

def gamma_correction(L: np.ndarray, gamma: float = 1.0) -> np.ndarray:
    """
    Corrección gamma.
    gamma < 1: Aclara (expande sombras)
    gamma > 1: Oscurece (comprime luces)
    """
    L = ensure_uint8(L)
    inv_gamma = 1.0 / gamma
    table = np.array([((i / 255.0) ** inv_gamma) * 255 
                      for i in range(256)]).astype(np.uint8)
    return cv2.LUT(L, table)


def adaptive_gamma(L: np.ndarray) -> np.ndarray:
    """Gamma adaptativo basado en luminosidad media."""
    L = ensure_uint8(L)
    mean_L = np.mean(L) / 255.0
    if mean_L > 0:
        gamma = -0.3 / np.log10(mean_L + 1e-6)
        gamma = np.clip(gamma, 0.5, 2.5)  # Limitar rango
    else:
        gamma = 1.0
    return gamma_correction(L, gamma)


# ============ FUNCIONES SIGMOIDALES ============

def sigmoid_contrast(L: np.ndarray, gain: float = 10, 
                     cutoff: float = 0.5) -> np.ndarray:
    """
    Contraste sigmoidal.
    gain: 5-20 (intensidad)
    cutoff: 0.3-0.7 (punto central)
    """
    L = ensure_uint8(L)
    L_norm = L / 255.0
    L_sigmoid = 1 / (1 + np.exp(-gain * (L_norm - cutoff)))
    # Renormalizar
    L_sigmoid = (L_sigmoid - L_sigmoid.min()) / (L_sigmoid.max() - L_sigmoid.min())
    return (L_sigmoid * 255).astype(np.uint8)


def logistic_contrast(L: np.ndarray, midpoint: int = 127, 
                      steepness: float = 0.05) -> np.ndarray:
    """
    Transformación logística.
    midpoint: valor central (0-255)
    steepness: pendiente (0.01-0.1)
    """
    L = ensure_uint8(L)
    L_logistic = 255 / (1 + np.exp(-steepness * (L - midpoint)))
    return L_logistic.astype(np.uint8)


# ============ TRANSFORMACIONES LOGARÍTMICAS ============

def log_transform(L: np.ndarray, c: float = 1.0) -> np.ndarray:
    """Transformación logarítmica - expande valores bajos."""
    L = ensure_uint8(L)
    L_log = c * np.log1p(L.astype(np.float32))
    L_log = (L_log / L_log.max() * 255)
    return L_log.astype(np.uint8)


def exp_transform(L: np.ndarray, c: float = 1.0) -> np.ndarray:
    """Transformación exponencial - expande valores altos."""
    L = ensure_uint8(L)
    L_norm = L / 255.0
    L_exp = np.expm1(c * L_norm)
    L_exp = (L_exp / L_exp.max() * 255)
    return L_exp.astype(np.uint8)


# ============ TRANSFORMACIONES ADAPTATIVAS ============

def unsharp_mask(L: np.ndarray, kernel_size: Tuple[int, int] = (5, 5),
                 sigma: float = 1.0, amount: float = 1.5, 
                 threshold: int = 0) -> np.ndarray:
    """
    Unsharp masking - realce de bordes.
    amount: 1.0-2.0 (intensidad)
    """
    L = ensure_uint8(L)
    blurred = cv2.GaussianBlur(L, kernel_size, sigma)
    sharpened = cv2.addWeighted(L, 1.0 + amount, blurred, -amount, 0)
    
    if threshold > 0:
        low_contrast_mask = np.absolute(L - blurred) < threshold
        np.copyto(sharpened, L, where=low_contrast_mask)
    
    return np.clip(sharpened, 0, 255).astype(np.uint8)


def retinex_single_scale(L: np.ndarray, sigma: float = 300) -> np.ndarray:
    """
    Single Scale Retinex.
    sigma: 100-300 (escala espacial)
    """
    L = ensure_uint8(L)
    L_float = L.astype(np.float32) + 1.0
    blurred = cv2.GaussianBlur(L_float, (0, 0), sigma)
    retinex = np.log10(L_float) - np.log10(blurred)
    
    # Normalizar
    retinex = retinex - retinex.min()
    retinex = (retinex / retinex.max() * 255)
    return retinex.astype(np.uint8)


def multi_scale_retinex(L: np.ndarray, 
                        sigmas: list = [15, 80, 250]) -> np.ndarray:
    """Multi-Scale Retinex."""
    L = ensure_uint8(L)
    L_float = L.astype(np.float32) + 1.0
    retinex = np.zeros_like(L_float)
    
    for sigma in sigmas:
        blurred = cv2.GaussianBlur(L_float, (0, 0), sigma)
        retinex += np.log10(L_float) - np.log10(blurred)
    
    retinex = retinex / len(sigmas)
    
    # Normalizar
    retinex = retinex - retinex.min()
    retinex = (retinex / retinex.max() * 255)
    return retinex.astype(np.uint8)


# ============ TRANSFORMACIONES MORFOLÓGICAS ============

def tophat_transform(L: np.ndarray, kernel_size: int = 15) -> np.ndarray:
    """Top-hat - realza objetos brillantes."""
    L = ensure_uint8(L)
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (kernel_size, kernel_size))
    tophat = cv2.morphologyEx(L, cv2.MORPH_TOPHAT, kernel)
    return cv2.add(L, tophat)


def blackhat_transform(L: np.ndarray, kernel_size: int = 15) -> np.ndarray:
    """Black-hat - realza objetos oscuros."""
    L = ensure_uint8(L)
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (kernel_size, kernel_size))
    blackhat = cv2.morphologyEx(L, cv2.MORPH_BLACKHAT, kernel)
    return cv2.subtract(L, blackhat)


# ============ MÉTODOS COMBINADOS ============

def clahe_plus_stretch(L: np.ndarray) -> np.ndarray:
    """CLAHE + Contrast stretching."""
    L_clahe = clahe_enhancement(L, clip_limit=2.0)
    return contrast_stretch(L_clahe, percentile=(2, 98))


def gamma_plus_unsharp(L: np.ndarray, gamma: float = 0.7) -> np.ndarray:
    """Gamma correction + Unsharp masking."""
    L_gamma = gamma_correction(L, gamma)
    return unsharp_mask(L_gamma, amount=1.5)


# ============ FUNCIÓN DE COMPARACIÓN PRINCIPAL ============

def compare_contrast_methods(L: np.ndarray, 
                            save_path: Optional[str] = None,
                            show_histograms: bool = True,
                            figsize: Tuple[int, int] = (20, 16)) -> Dict[str, np.ndarray]:
    """
    Compara todos los métodos de mejora de contraste.
    
    Args:
        L: Canal L de la imagen (grayscale, 0-255)
        save_path: Si se proporciona, guarda la figura (ej: 'comparison.png')
        show_histograms: Si True, muestra histogramas sobrepuestos
        figsize: Tamaño de la figura
    
    Returns:
        Diccionario con todas las imágenes procesadas
    """
    
    # Asegurar que L esté en uint8
    L = ensure_uint8(L)
    
    # Aplicar todos los métodos
    methods = {
        '1. Original': L,
        '2. Contrast Stretch': contrast_stretch(L),
        '3. Linear Scale (α=1.5)': linear_scaling(L, alpha=1.5),
        '4. Histogram Eq.': histogram_equalization(L),
        '5. CLAHE': clahe_enhancement(L, clip_limit=2.0),
        '6. Gamma 0.5 (Bright)': gamma_correction(L, gamma=0.5),
        '7. Gamma 1.5 (Dark)': gamma_correction(L, gamma=1.5),
        '8. Adaptive Gamma': adaptive_gamma(L),
        '9. Sigmoid': sigmoid_contrast(L, gain=10),
        '10. Logistic': logistic_contrast(L, steepness=0.05),
        '11. Log Transform': log_transform(L),
        '12. Exp Transform': exp_transform(L),
        '13. Unsharp Mask': unsharp_mask(L, amount=1.5),
        '14. Retinex (Single)': retinex_single_scale(L, sigma=200),
        '15. Retinex (Multi)': multi_scale_retinex(L),
        '16. Top-Hat': tophat_transform(L, kernel_size=15),
        '17. Black-Hat': blackhat_transform(L, kernel_size=15),
        '18. CLAHE + Stretch': clahe_plus_stretch(L),
        '19. Gamma + Unsharp': gamma_plus_unsharp(L, gamma=0.7),
    }
    
    # Calcular estadísticas
    print("=" * 80)
    print("COMPARACIÓN DE MÉTODOS DE CONTRASTE")
    print("=" * 80)
    print(f"{'Método':<25} {'Mean':<8} {'Std':<8} {'Min':<6} {'Max':<6} {'Range':<6}")
    print("-" * 80)
    
    for name, img in methods.items():
        mean_val = np.mean(img)
        std_val = np.std(img)
        min_val = np.min(img)
        max_val = np.max(img)
        range_val = max_val - min_val
        print(f"{name:<25} {mean_val:>7.2f} {std_val:>7.2f} {min_val:>5} {max_val:>5} {range_val:>5}")
    
    print("=" * 80)
    
    # Crear figura
    n_methods = len(methods)
    n_cols = 5
    n_rows = int(np.ceil(n_methods / n_cols))
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize)
    axes = axes.ravel()
    
    # Plotear cada método
    for idx, (name, img) in enumerate(methods.items()):
        ax = axes[idx]
        
        # Mostrar imagen
        ax.imshow(img, cmap='gray', vmin=0, vmax=255)
        ax.set_title(name, fontsize=10, fontweight='bold')
        ax.axis('off')
        
        # Agregar histograma si se solicita
        if show_histograms:
            # Crear eje secundario para histograma
            ax_hist = ax.inset_axes([0.05, 0.05, 0.9, 0.25])
            ax_hist.hist(img.ravel(), bins=64, range=(0, 256), 
                        color='cyan', alpha=0.6, edgecolor='none')
            ax_hist.set_xlim(0, 255)
            ax_hist.set_xticks([])
            ax_hist.set_yticks([])
            ax_hist.patch.set_alpha(0.3)
            
            # Estadísticas en la imagen
            stats_text = f"μ={np.mean(img):.1f} σ={np.std(img):.1f}"
            ax.text(0.5, 0.95, stats_text, 
                   transform=ax.transAxes,
                   ha='center', va='top',
                   fontsize=8,
                   bbox=dict(boxstyle='round', facecolor='black', alpha=0.5),
                   color='white')
    
    # Ocultar ejes sobrantes
    for idx in range(n_methods, len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    
    # Guardar si se especifica
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"\n✓ Figura guardada en: {save_path}")
    
    plt.show()
    
    return methods


# ============ FUNCIONES DE COMPARACIÓN ESPECÍFICAS ============

def compare_gamma_range(L: np.ndarray, 
                       gammas: list = [0.3, 0.5, 0.7, 1.0, 1.3, 1.7, 2.0, 2.5],
                       save_path: Optional[str] = None) -> Dict[str, np.ndarray]:
    """Compara diferentes valores de gamma."""
    
    methods = {'Original': L}
    for gamma in gammas:
        methods[f'Gamma {gamma}'] = gamma_correction(L, gamma)
    
    n_methods = len(methods)
    n_cols = 3
    n_rows = int(np.ceil(n_methods / n_cols))
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, 5 * n_rows))
    axes = axes.ravel()
    
    for idx, (name, img) in enumerate(methods.items()):
        axes[idx].imshow(img, cmap='gray')
        axes[idx].set_title(name, fontsize=12, fontweight='bold')
        axes[idx].axis('off')
        
        # Histograma
        axes[idx].hist(img.ravel(), bins=64, alpha=0.3, color='blue')
    
    for idx in range(n_methods, len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"✓ Guardado: {save_path}")
    
    plt.show()
    
    return methods


def compare_clahe_params(L: np.ndarray,
                        clip_limits: list = [1.0, 2.0, 3.0, 4.0],
                        tile_sizes: list = [(4,4), (8,8), (16,16)],
                        save_path: Optional[str] = None) -> Dict[str, np.ndarray]:
    """Compara diferentes parámetros de CLAHE."""
    
    methods = {'Original': L}
    
    for clip in clip_limits:
        for tile in tile_sizes:
            name = f'CLAHE clip={clip} tile={tile}'
            methods[name] = clahe_enhancement(L, clip_limit=clip, tile_size=tile)
    
    n_methods = len(methods)
    n_cols = 4
    n_rows = int(np.ceil(n_methods / n_cols))
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
    axes = axes.ravel()
    
    for idx, (name, img) in enumerate(methods.items()):
        axes[idx].imshow(img, cmap='gray')
        axes[idx].set_title(name, fontsize=9, fontweight='bold')
        axes[idx].axis('off')
    
    for idx in range(n_methods, len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"✓ Guardado: {save_path}")
    
    plt.show()
    
    return methods

```

```python
def _ensure_uint8(L: np.ndarray) -> np.ndarray:
    """Asegura que la imagen esté en formato uint8."""
    if L.dtype != np.uint8:
        # Si está en float [0, 1], escalar a [0, 255]
        if L.max() <= 1.0:
            L = (L * 255).astype(np.uint8)
        else:
            # Si está en otro rango, clip y convertir
            L = np.clip(L, 0, 255).astype(np.uint8)
    return L

def gamma_contrast(L: np.ndarray, 
                     gamma=1.0, 
                     plot = False):
    """
    gamma < 1: Aclara sombras (expande bajos valores)
    gamma > 1: Oscurece luces (comprime altos valores)
    gamma = 1: Sin cambio
    """

    # Normalizar a [0, 1]
    L = _ensure_uint8(L)
    L_norm = L / 255.0
    # Aplicar gamma
    L_corrected = np.power(L_norm, gamma)
    # Volver a [0, 255]
    l_gamma_corrected = (L_corrected * 255).astype(np.uint8)
    if plot:
        plot_img(l_gamma_corrected)

    return l_gamma_corrected

def sigmoid_contrast(L: np.ndarray, gain: float = 10, 
                     cutoff: float = 0.5) -> np.ndarray:
    """
    Contraste sigmoidal.
    gain: 5-20 (intensidad)
    cutoff: 0.3-0.7 (punto central)
    """
    L = _ensure_uint8(L)
    L_norm = L / 255.0
    L_sigmoid = 1 / (1 + np.exp(-gain * (L_norm - cutoff)))
    # Renormalizar
    L_sigmoid = (L_sigmoid - L_sigmoid.min()) / (L_sigmoid.max() - L_sigmoid.min())
    return (L_sigmoid * 255).astype(np.uint8)

def exp_transform(L: np.ndarray, c: float = 1.0) -> np.ndarray:
    """Transformación exponencial - expande valores altos."""
    L = _ensure_uint8(L)
    L_norm = L / 255.0
    L_exp = np.expm1(c * L_norm)
    L_exp = (L_exp / L_exp.max() * 255)
    return L_exp.astype(np.uint8)


def apply_contrast(img: np.ndarray, 
                   contrast_method: Optional[str] = 'gamma',
                   gamma: Optional[float] = 1.5,
                   gain: Optional[float] = 5,
                   cutoff: Optional[float] = 0.5,
                   c: Optional[float] = 0.5,
                   plot: Optional[bool] = False,
                   plot_size: Optional[bool] = False,
                   compare: Optional[bool] = False) -> np.ndarray:
    """
    Aplica transformación de contraste al canal L de una imagen LAB.
    
    Args:
        img: Imagen en formato BGR
        contrast_method: 'gamma', 'sigmoid', o 'exp'
        gamma: Parámetro para gamma_contrast (default: 1.0)
        gain: Parámetro para sigmoid_contrast (default: 10)
        cutoff: Parámetro para sigmoid_contrast (default: 0.5)
        c: Parámetro para exp_transform (default: 1.0)
        plot: Si True, muestra el resultado del método seleccionado
        compare: Si True, compara visualmente los 3 métodos (ignora plot)
    
    Returns:
        Canal L transformado (numpy array 2D)
    """
    # Validar entrada
    if not isinstance(img, np.ndarray):
        raise TypeError("Input image must be a numpy array")
    if img.ndim != 3 or img.shape[2] != 3:
        raise ValueError("Image must be in BGR format (3 channels)")
    
    # Convertir a LAB
    lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    
    # Extraer canal L
    l_channel = lab[:, :, 0]
    
    # Diccionario de métodos
    contrast_methods = {
        'gamma': lambda: gamma_contrast(l_channel, gamma=gamma, plot=False),
        'sigmoid': lambda: sigmoid_contrast(l_channel, gain=gain, cutoff=cutoff),
        'exp': lambda: exp_transform(l_channel, c=c)
    }
    
    # Validar método
    if contrast_method not in contrast_methods:
        raise ValueError(f"contrast_method must be one of {list(contrast_methods.keys())}")
    
    # Si compare=True, mostrar los 3 métodos
    if compare:
        l_gamma = gamma_contrast(l_channel, gamma=gamma, plot=False)
        l_sigmoid = sigmoid_contrast(l_channel, gain=gain, cutoff=cutoff)
        l_exp = exp_transform(l_channel, c=c)
        
        plt.figure(figsize=plot_size)
        
        plt.subplot(2, 2, 1)
        plt.imshow(l_channel, cmap='gray')
        plt.title("Original")
        plt.axis('off')
        
        plt.subplot(2, 2, 2)
        plt.imshow(l_gamma, cmap='gray')
        plt.title(f"Gamma (γ={gamma})")
        plt.axis('off')
        
        plt.subplot(2, 2, 3)
        plt.imshow(l_sigmoid, cmap='gray')
        plt.title(f"Sigmoid (gain={gain}, cutoff={cutoff})")
        plt.axis('off')
        
        plt.subplot(2, 2, 4)
        plt.imshow(l_exp, cmap='gray')
        plt.title(f"Exponencial (c={c})")
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
    
    # Aplicar transformación al canal L del método seleccionado
    l_transformed = contrast_methods[contrast_method]()
    
    # Si plot=True y compare=False, mostrar solo el método seleccionado
    if plot and not compare:
        plt.figure(figsize=plot_size)
        
        plt.subplot(1, 2, 1)
        plt.imshow(l_channel, cmap='gray')
        plt.title("Canal L Original")
        plt.axis('off')
        
        plt.subplot(1, 2, 2)
        plt.imshow(l_transformed, cmap='gray')
        plt.title(f"Canal L Transformado ({contrast_method})")
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
    
    return l_transformed
```
